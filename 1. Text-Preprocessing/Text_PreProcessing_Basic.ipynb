#4. Remove Punctuations

```python
import time
import string

exclude=string.punctuation
```

```python
exclude
```

```python
def remove_punch(text):
  for char in exclude:
    text=text.replace(char, ' ')
  return text
```

```python
def chat_conversion(text):
    new_text = []
    for w in text.split():
        if w.upper() in chat_words_map: # Use chat_words_map
            new_text.append(chat_words_map[w.upper()])
        else:
            new_text.append(w)
    return " ".join(new_text)

# Apply the simplified chat word conversion to the 'review' column
dataset['review'] = dataset['review'].apply(chat_conversion)

print("Chat word conversion applied to the 'review' column.")
print(dataset['review'].head())
```

## 7. Removing Stop Words

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
remove=stopwords.words('english')
```

```python
remove
```

```python
def reomve_stopwords(text):
  new_text=[]

  for word in text.split():
    if word in remove:
      new_text.append(' ')
    else:
      new_text.append(word)

  x=new_text[:]
  new_text.clear()
  return " ".join(x)
```

```python
dataset.review
```

```python
dataset['review'] = dataset['review'].apply(reomve_stopwords)
```

## 8. Handling Emojis

```python
# For some application either we have to remove it or replace it totally depends upon the application
```

```python
import re
def remove_emoji(text):
    emoji_pattern = re.compile("["
                                u"\U0001F600-\U0001F64F"  # emoticons
                                u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                                u"\U0001F680-\U0001F6FF"  # transport & map symbols
                                u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                                u"\U00002702-\U000027B0"
                                u"\U000024C2-\U0001F251"
                                "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps=PorterStemmer()
def stem_words(text):
  return " ".join(ps.stem(word) for word in text.split())
```

### Lemitization

```python
import nltk
nltk.download('wordnet') # Added to download the wordnet corpus
from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer = WordNetLemmatizer()

sentence = "He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
punctuations="?:!.,;"
sentence_words = nltk.word_tokenize(sentence)
for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)

sentence_words
print("{0:20}{1:20}".format("Word", "Lemma"))
for word in sentence_words:
    print ("{0:20}{1:20}".format(word, wordnet_lemmatizer.lemmatize(word,pos= 'v')))
```
