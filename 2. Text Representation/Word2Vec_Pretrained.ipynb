```python
print("Words most similar to 'king - man + woman' (e.g., find the 'queen' in this analogy):")
print(word_vectors.most_similar(positive=['king', 'woman'], negative=['man']))


import numpy as np # Ensure numpy is imported for vector operations

analogy_vector = word_vectors['king'] - word_vectors['man'] + word_vectors['woman']
man_vector = word_vectors['queen']


cosine_sim = np.dot(man_vector, analogy_vector) / (np.linalg.norm(man_vector) * np.linalg.norm(analogy_vector))

print(f"\nCosine similarity between 'man' and the analogy vector (king - man + woman): {cosine_sim}")
```
